```python
import pandas as pd
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
np.random.seed(42)
years = list(range(1991, 2026))

df = pd.DataFrame({
    'Year': years,
    'T': np.round(np.random.uniform(12, 17, len(years)), 1),
    'TM': np.round(np.random.uniform(19, 24, len(years)), 1),
    'Tm': np.round(np.random.uniform(6, 11, len(years)), 1),
    'V': np.round(np.random.uniform(4, 11, len(years)), 1),
    'RA': np.random.randint(30, 120, len(years)),
    'SN': np.random.randint(5, 40, len(years)),
    'TS': np.random.randint(0, 20, len(years)),
    'FG': np.random.randint(0, 35, len(years)),
    'TN': np.random.randint(0, 3, len(years)),
    'GR': np.random.randint(0, 3, len(years)),
})

df.set_index('Year', inplace=True)
df.to_csv('df.csv')
df
```

```python
df=pd.read_csv('df.csv',index_col="Year")
df
```

```python
df.plot()
plt.show()
```

```python
df.plot()
plt.title("Yillik ob-havo ko'rsatkichlari")
plt.show()
```

```python
df.plot()
plt.title("Yillik ob-havo ko'rsatkichlari")
plt.xlabel("Yil")
plt.ylabel("Ko'rsatkichlar")
plt.show()
```

```python
df[['T', 'TM', 'Tm']].plot()
plt.title("Yillik ob-havo ko'rsatkichlari")
plt.xlabel("Yil")
plt.ylabel("Ko'rsatkichlar")
plt.show()

```

```python
df[['RA', 'SN', 'TS', 'FG', 'GR']].plot()
plt.title("Yillik ob-havo ko'rsatkichlari")
plt.xlabel("Yil")
plt.ylabel("Ko'rsatkichlar")
plt.show()
```

```python
sns.barplot(x=df.index, y=df['V'])
plt.title("Yillik o'rtacha shamol tezligi")
plt.xlabel("Yil")
plt.ylabel("Shamol tezligi")
plt.grid(color="gray", alpha=0.3)
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.heatmap(df[['RA','SN','TS','FG','GR']],cmap='Accent',linewidths=0,square=True)
plt.show()
```
