```python
import pandas as pd
import numpy as np
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("https://github.com/anvarnarz/praktikum_datasets/raw/main/museum_visitors.csv", index_col=0)
df.head()
```

```python
df.shape
```

```python
sns.lineplot(data=df)
plt.show()
```

```python
plt.figure(figsize=(16,6))        # Grafikning o‘lchamini belgilaydi: eni 16, bo‘yi 6

plt.title("Muzeylarga tashriflar soni")  # Grafikning sarlavhasini beradi

plt.xlabel("Yil")                 # X o‘qiga "Yil" nomini beradi

plt.ylabel("Odamlar soni")        # Y o‘qiga "Odamlar soni" nomini beradi

plt.xticks(rotation=90)           # X o‘qidagi yozuvlarni 90° buradi

sns.lineplot(data=df)             # df ma’lumotlari asosida chiziqli grafik chizadi

plt.show()                        # Tayyor grafikni ekranga chiqaradi
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/anvarnarz/praktikum_datasets/main/uzbekistan.csv", index_col=0)
df.columns = ['maydon','aholi']
df
```

```python
plt.figure(figsize=(16,6))
sns.barplot(x=df.index,y = df['maydon'])
plt.title("O‘zbekiston Respublikasi hududi aholi soni")
plt.xlabel("Hudud")
plt.ylabel("Aholi soni(million)")
plt.grid()
plt.xticks(rotation=90)
plt.show()
```

```python
import pandas as pd

data = {
    'mahsulot': ['Telefon', 'Noutbuk', 'Naushnik', 'Sichqoncha', 'Klaviatura', 'Monitor'],
    'kategoriya': ['Elektronika', 'Elektronika', 'Aksessuar', 'Aksessuar', 'Aksessuar', 'Elektronika'],
    'narx': [3500000, 8500000, 250000, 120000, 300000, 1800000],
    'sotildi': [45, 12, 80, 60, 55, 20]
}

df = pd.DataFrame(data)
df
```

```python
#plt.figure(figsize = (16,6))
sns.barplot(x=df["mahsulot"],y=df["sotildi"])
plt.title("Mahsulotlarning sotilganlari")
plt.xlabel("mahsulotlar")
plt.ylabel("Sotilganlar")
plt.grid(color="gray", alpha=0.3)

plt.show()
```

```python
#plt.figure(figsize = (16,6))
sns.lineplot(x=df["mahsulot"],y=df["sotildi"])
plt.title("Mahsulotlarning sotilganligi")
plt.xlabel("mahsulotlar",color = 'r')
plt.ylabel("Sotilganlar",color = 'b')
plt.grid(color="gray", alpha=0.3)
plt.show()
```
