```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("https://github.com/anvarnarz/praktikum_datasets/raw/main/automobile_data_nan.csv",index_col=0)
df.head()
```

```python
df.isnull()
```

```python
df.shape
```

```python
df.fillna(0)
```

```python
df.fillna({'wheel-base':100})
```

```python
df.fillna({'wheel-base':100,'engine-type':99})
```

```python
df.fillna({'wheel-base':df['wheel-base'].mean()})
```

```python
df.fillna(method='ffill')
```

```python
df[df.company=='toyota']
```

```python
df[df.company=='toyota'].fillna(method='bfill')
```

```python
df2 =df.fillna(1)
df2
```

```python
dfh = df[df['body-style']=='hatchback']
dfh
```

```python
dfh.fillna(dfh.mean(numeric_only=True))
```

```python
dfh = dfh.fillna({'engine-type': dfh['engine-type'].mode()[0]})
dfh
```
