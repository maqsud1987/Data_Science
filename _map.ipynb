```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("https://github.com/anvarnarz/praktikum_datasets/raw/main/automobile_data_nan.csv",index_col=0)
df
```

```python
df.shape
```

```python
df['num-of-cylinders'].unique()
```

```python
dict={'two':2,'three':3,'four':4,'five':5,'six':6,'eight':8,'twelve':12}
df['cylinders']=df['num-of-cylinders'].map(dict)
df.head()
```

```python
df['price']
```

```python
kurs=10720
df['price_uz']=df['price'].map(lambda x: x*kurs)
df.head()
```

```python
fuel_dict = {'gas': 1, 'diesel': 2}
df['fuel-num']=df['fuel-type'].map(fuel_dict)
df
```

```python
dict = {'two':2, 'four':4}
df['doors_num'] = df['num-of-doors'].map(dict)
```
