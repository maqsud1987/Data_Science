```python
import pandas as pd
data = {
    'apples': [3, 2, 0, 1],
    'oranges': [0, 3, 7, 2]
}
df = pd.DataFrame(data)
#print(df)    #  ro'yxat to'liq chiqadi
df.head()    #  ro'yxat 2 qator chiqadi
data

```

```python
df = pd.DataFrame(data,columns=['apples','oranges','narx','sotildi'])
df  # columns - DataFrame parametr, ustunlarga nom berish uchun.
```

```python
df = pd.DataFrame(data,columns=['apples','oranges','narx','sotildi'],index=['a','b','c','d'])
df
df
```

```python
df.sotildi = 0
df
```

```python
df[['sotildi','narx']] = 0 # 2 ta ustinni bir vaqtda 0 qilish
df
```
